# AI_Polymorphism_Detection_Sim Analysis
This notebook loads the combined dataset CSV and shows simple feature explorations on synthetic polymorphic PE artifacts.

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv('../dataset/combined_features.csv')
df.head()
```

```python
df['label'].value_counts().plot(kind='bar', title='Label Distribution')
plt.ylabel('Count')
plt.show()
```

```python
features = ['num_sections', 'avg_section_entropy', 'imported_api_count', 'strings_count', 'junk_bytes_count', 'resource_size']
df[features].hist(figsize=(10, 8), bins=20)
plt.tight_layout()
plt.show()
```
